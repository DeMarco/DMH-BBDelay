```python
import scipy
import numpy as np
import matplotlib.pyplot as plt

plt.style.use(['science', 'notebook', 'grid'])
```

```python
from scipy.special import lambertw
```

Original resistor values:

- R1 = 10k
- R2 = 62k
- R3 = 10k
- R4 = 90.9k
- R5 = 100k
- R6 = 200k
- R7 = 200k

```python
# Original resistors:

R1_o = 10000
R2_o = 62000
R3_o = 10000
R4_o = 90900
R5_o = 100000
R6_o = 200000
R7_o = 200000

# New resistors

R1_n = 10000
R2_n = 62000
R3_n = 10000
R4_n = 90900
R5_n = 100000
R6_n = 100000
R7_n = 100000
```

```python
Vin = np.linspace(0.00007746,8,100000)

# Calculate graph with original resistors:

u_o = (3/2)*(R7_o/R5_o)*(R4_o/R3_o)*(1/np.sqrt(2))*np.log(10)

print(u_o)

Vout_o = lambertw(u_o*(R2_o/R1_o)*Vin)/u_o

# Calculate graph with new resistors:

u_n = (3/2)*(R7_n/R5_n)*(R4_n/R3_n)*(1/np.sqrt(2))*np.log(10)

print(u_n)

Vout_n = lambertw(u_n*(R2_n/R1_n)*Vin)/u_n

Vin_p = Vin * np.sqrt(2)
Vout_p_o = Vout_o * np.sqrt(2)
Vout_p_n = Vout_n * np.sqrt(2)

fig1 = plt.figure()
ax = fig1.add_subplot(1, 1, 1)
ax.plot(Vin_p, Vout_p_o, '-.', color='purple')
ax.plot(Vin_p, Vout_p_n, color='green')
ax.set_xlim(0, 11)
ax.set_ylim(0, 0.5)
plt.xlabel('Vin [Vp]')
plt.ylabel('Vout [Vp]')

Vin_pp = Vin_p * 2
Vout_pp_o = Vout_p_o * 2
Vout_pp_n = Vout_p_n * 2

fig2 = plt.figure()
ax = fig2.add_subplot(1, 1, 1)
ax.plot(Vin_pp, Vout_pp_o, '-.', color='purple')
ax.plot(Vin_pp, Vout_pp_n, color='green')
ax.set_xlim(0, 11)
ax.set_ylim(0, 1)
plt.xlabel('Vin [Vpp]')
plt.ylabel('Vout [Vpp]')

V0 = 0.7746

Vin_dBu = 20*(np.log10(Vin/V0))
Vout_dBu_o = 20*(np.log10(Vout_o/V0))
Vout_dBu_n = 20*(np.log10(Vout_n/V0))

fig3 = plt.figure()
ax = fig3.add_subplot(1, 1, 1)
ax.plot(Vin_dBu, Vin_dBu, '--', color='blue')
ax.plot(Vin_dBu, Vout_dBu_o, '-.', color='purple')
ax.plot(Vin_dBu, Vout_dBu_n, '-', color='green')
plt.xlabel('Vin [dBu]')
plt.ylabel('Vout [dBu]')
```
